# Сравнение моделей

```python
from clearml import Task
import pandas as pd
```

```python
# Подключаемся к проекту
project_name = "MLOps project"

xgb_task_name = "xgb task"
catboost_opt_task_name = "catboost optimized (Optuna)"
catboost_task_name = "catboost task"
```

```python
def get_metrics_artifact(project_name, task_name, artifact='metrics'):
    """ Получаем задачу по имени проекта и имени задачи """
    task = Task.get_task(project_name=project_name, task_name=task_name)
    if task:
        metrics_artifact = task.artifacts.get(artifact)
        if metrics_artifact:
            metrics_json = metrics_artifact.get()
            return metrics_json
    return None
```

```python
xgb_metrics = get_metrics_artifact(project_name, xgb_task_name)
cb_metrics = get_metrics_artifact(project_name, catboost_task_name)
cbo_metrics = get_metrics_artifact(project_name, catboost_opt_task_name)
```

```python
xgb_metrics
```

```python

metrics_results = {
    'xgb': xgb_metrics,
    'catboost opt': cb_metrics,
    'catboost': cbo_metrics
}

df_metrics = pd.DataFrame.from_dict(metrics_results, orient='index')

df_metrics.index.name = 'Model'
df_metrics = df_metrics.reset_index()

df_metrics
```

По метрикам видно, что тюненый catboost близок к xgb.
